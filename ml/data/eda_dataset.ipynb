```python
import pandas as pd

# load your 40K dataset
df = pd.read_csv("raw/real-world.csv")  # adjust name if needed

# rename column
df.rename(columns={"text_": "review"}, inplace=True)

df.head()
```

```python
# normalize labels
df["label"] = df["label"].str.lower().str.strip()

# keep only valid labels
df = df[df["label"].isin(["fake", "genuine"])]

df["label"].value_counts()
```

```python
before = len(df)

df.drop_duplicates(subset=["review"], inplace=True)

after = len(df)

print(f"Removed {before - after} duplicates")
```

```python
df["label"].value_counts()
```

```python
fake_df = df[df["label"] == "fake"]
genuine_df = df[df["label"] == "genuine"]

# balance
genuine_df = genuine_df.sample(len(fake_df), random_state=42)

df = pd.concat([fake_df, genuine_df])

df["label"].value_counts()
```

```python
df = df.sample(3000, random_state=42)

print("Final size:", len(df))
```

```python
df["length"] = df["review"].apply(lambda x: len(str(x).split()))

hard_df = df[
    (df["length"] < 5) |
    (df["length"] > 25)
]

print("Hard samples:", len(hard_df))
```

```python
df[["review", "label"]].to_csv("raw/filtered_40k.csv", index=False)

# optional (with rating)
df.to_csv("raw/filtered_40k_with_rating.csv", index=False)

print("Saved successfully")
```
