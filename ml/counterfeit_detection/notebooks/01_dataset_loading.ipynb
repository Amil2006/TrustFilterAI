```python
import os
import pandas as pd
```

```python
# Path setup
BASE_PATH = os.path.abspath(os.path.join(os.getcwd(), ".."))
DATASET_PATH = os.path.join(BASE_PATH, "data", "raw")

print("Dataset Path:", DATASET_PATH)
```

```python
# Load dataset 
classes = ["genuine", "fake"]
data = []

for brand in os.listdir(DATASET_PATH):
    brand_path = os.path.join(DATASET_PATH, brand)
    
    if not os.path.isdir(brand_path):
        continue
    
    for label in classes:
        label_path = os.path.join(brand_path, label)
        
        if not os.path.exists(label_path):
            continue
        
        for img_name in os.listdir(label_path):
            img_path = os.path.join(label_path, img_name)
            
            data.append({
                "image_path": img_path,
                "label": label,
                "brand": brand
            })

df = pd.DataFrame(data)
df.head()
```

```python
print("Total Images:", len(df))

print("\nClass Distribution:")
print(df["label"].value_counts())

print("\nBrand Distribution:")
print(df["brand"].value_counts())
```

```python
# Map labels to numeric values
label_map = {"genuine": 0, "fake": 1}
df["label_num"] = df["label"].map(label_map)

df.head()
```

```python
# Save the processed dataset to a CSV file
output_path = os.path.join(BASE_PATH, "data", "processed", "dataset.csv")

os.makedirs(os.path.dirname(output_path), exist_ok=True)

df.to_csv(output_path, index=False)

print("Saved dataset to:", output_path)
```
